```python


# Initialize client
import os
import json
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage
from textwrap import indent
from dotenv import load_dotenv
# Load environment variables
load_dotenv()

import warnings
warnings.filterwarnings('ignore')
```

```python
# Configure llm via .env

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
```

```python
def compare_prompts(raw_prompt: str, tuned_prompt: str, title="Comparison"):
    """
    Compare raw prompt vs tuned prompt using the same LLM.
    Prints clean side-by-side responses.
    """

    print(f"{title}")
    print("==============================\n")

    # RAW PROMPT CALL
    raw_messages = [
        SystemMessage(content="You are a helpful assistant."),
        HumanMessage(content=raw_prompt)
    ]
    raw_response = llm(raw_messages).content.strip()

    # TUNED PROMPT CALL
    tuned_messages = [
        SystemMessage(content="You are a helpful assistant."),
        HumanMessage(content=tuned_prompt)
    ]
    tuned_response = llm(tuned_messages).content.strip()

    # PRINT OUTPUTS
    print("RAW PROMPT:")
    print("--------------------------------")
    print(raw_prompt)
    print("\nRAW RESPONSE:")
    print(indent(raw_response, "    "))

    print("\n\nTUNED PROMPT:")
    print("--------------------------------")
    print(tuned_prompt)
    print("\nTUNED RESPONSE:")
    print(indent(tuned_response, "    "))

    print("\n==============================")
    print("END OF COMPARISON")
```

```python
# EXAMPLE USAGE — FINTECH LOAN DECISION EXPLAINER

raw_prompt_loan = """
Explain why the customer was denied the loan.
"""

tuned_prompt_loan = """
You are a financial assistant that explains loan decisions clearly.

Rules:
• Do NOT invent reasons — only use provided ones.
• Do NOT mention internal scoring systems.
• Be supportive and simple.
• Provide next steps.

Input:
Loan Status: DENIED
Reason Provided: Low credit history, missing income documents.

Format:
1. Summary
2. Main Factors (bullets)
3. Next Steps
"""

compare_prompts(raw_prompt_loan, tuned_prompt_loan, title="FinTech Loan Decision Explainer")
```

## SUPPORT TICKET SUMMARIZATION

```python
# ------------------------------------------------------------
# EXAMPLE USAGE — SUPPORT TICKET SUMMARIZATION
# ------------------------------------------------------------

raw_prompt_ticket = """
Summarize this customer issue:
User says app isn’t loading, tried reinstalling, still breaks on login.
"""

tuned_prompt_ticket = """
You are a support assistant. Convert ticket logs to structured summaries.

Rules:
• Identify main issue.
• Extract troubleshooting attempts.
• Capture impact.
• Provide agent next steps.
• Produce a short customer-facing message.

Output:
Issue Summary:
Attempts Already Made:
Impact:
Recommended Agent Actions:
Customer-Facing Message:
"""

compare_prompts(raw_prompt_ticket, tuned_prompt_ticket, title="Support Ticket Summarization")
```

```python
#RAW PROMPT (with input)
raw_prompt_email = """Reply to this customer email:

"I am extremely frustrated. My order was supposed to arrive last week.
No one is responding and this is unacceptable."
"""
#TUNED PROMPT (with input)
tuned_prompt_email = """
You are a Senior Customer Success Manager.

Tone:
- Calm
- Empathetic
- Professional

Goal:
1. Acknowledge frustration
2. Briefly explain the delay
3. Provide a clear next step

Constraints:
- No excuses
- No blaming
- Max 120 words

Customer email:
"I am extremely frustrated. My order was supposed to arrive last week.
No one is responding and this is unacceptable."
"""

compare_prompts(raw_prompt_email, tuned_prompt_email, title="Support Email Response")
```

```python
raw_prompt_vector_db = "Explain what a vector database is."

tuned_prompt_vector_db = """
You are explaining to a Product Manager with no machine learning background.

Context:
The listener understands products and user experience but not AI terminology.

Task:
Explain what a Vector Database is.

Instructions:
1. Use one real-life analogy
2. Use one popular app example (Google, Netflix, or Amazon)
3. Explain why it is useful, not how it is built

Constraints:
- No math
- No technical or ML jargon
- Simple, conversational language
- Maximum 120 words

Output format:
- Analogy
- App example
- Why it matters
"""

compare_prompts(raw_prompt_vector_db, tuned_prompt_vector_db, title="Vector DB explanation")
```

```python
over_tuned_prompt_vector_db = """
You are a globally renowned AI researcher, database architect, and product visionary with 25+ years of experience.

Context:
Assume the audience is a Product Manager, but also assume they can quickly learn advanced concepts if explained thoroughly.

Task:
Explain what a Vector Database is in extreme detail.

Instructions:
1. Start with a real-life analogy
2. Then explain mathematical foundations of vector embeddings
3. Describe cosine similarity, dot product, and Euclidean distance
4. Explain indexing techniques such as HNSW, IVF, PQ
5. Compare Vector Databases with Relational Databases and Graph Databases
6. Mention popular Vector DBs and their internal architectures

Constraints:
- Be very thorough
- Cover all edge cases
- Do not oversimplify
- Use correct technical terminology

Output format:
- Analogy
- Mathematical explanation
- Indexing mechanisms
- Database comparison table
- Architecture overview
- Summary
"""

compare_prompts(tuned_prompt_vector_db, over_tuned_prompt_vector_db, title="Vector DB explanation")
```
